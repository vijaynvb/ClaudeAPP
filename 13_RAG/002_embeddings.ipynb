# Text Embeddings

**Definition:** An embedding is a list of numbers (a vector) that represents the *meaning* of a piece of text. Texts with similar meaning get vectors that point in similar directions — this is what lets us search by meaning instead of by matching exact words.

We use Voyage AI's embedding model here (Claude itself doesn't generate embeddings) via the `voyageai` client. Turning a chunk into an embedding is the step that makes it searchable — this notebook shows just that one call.

```python
# Client setup — reads VOYAGE_API_KEY from the .env file next to this notebook
from dotenv import load_dotenv
import voyageai

load_dotenv()

client = voyageai.Client()
```

## Generating an embedding

`input_type` tells Voyage which side of a future search this text is on: `"document"` for text you're storing, `"query"` for a question you'll search with. The model encodes them slightly differently, and using the right one improves retrieval — we'll see both used in the next notebook.


```python
def generate_embedding(text, model="voyage-3-large", input_type="document"):
    result = client.embed([text], model=model, input_type=input_type)
    return result.embeddings[0]
```

```python
sample_text = "Compound CTX-204b showed promising results in its Phase IIa clinical trial."

embedding = generate_embedding(sample_text)

print(f"{len(embedding)} dimensions")
print("First 5 values:", embedding[:5])
```
